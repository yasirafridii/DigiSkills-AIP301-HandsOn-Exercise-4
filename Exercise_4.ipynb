```python
import urllib.request
import ssl

# SSL verification bypass karne ke liye
ctx = ssl.create_default_context()
ctx.check_hostname = False
ctx.verify_mode = ssl.CERT_NONE

# Mushrooms CSV file download
url = "https://lms.dl.digiskills.pk/Courses/AIP301/Downloads/mushrooms.csv" # agar URL mein koi masla ho toh neeche wala simple method try karein
req = urllib.request.urlopen("https://lms.digiskills.pk/Courses/AIP301/Downloads/mushrooms.csv", context=ctx)
with open("mushrooms.csv", "wb") as f:
    f.write(req.read())

print("Mushrooms CSV file successfully download ho gayi hai!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, SimpleRNN, Dense, Flatten
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

# Load the dataset
df = pd.read_csv('mushrooms.csv')

# Convert categorical features into numerical format
label_encoder = LabelEncoder()
for col in df.columns:
    df[col] = label_encoder.fit_transform(df[col])

# Split features and target
X = df.drop('class', axis=1).values
y = df['class'].values

# Split dataset into 70% training and 30% testing data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Reshape data for CNN and RNN input requirements
X_train_reshaped = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
X_test_reshaped = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

print("Data successfully preprocess ho gaya hai!")
```

```python
import urllib.request
import ssl

ctx = ssl.create_default_context()
ctx.check_hostname = False
ctx.verify_mode = ssl.CERT_NONE

req = urllib.request.urlopen("https://lms.digiskills.pk/Courses/AIP301/Downloads/exercise_text.json", context=ctx)
with open("exercise_text.json", "wb") as f:
    f.write(req.read())

print("exercise_text.json file successfully download ho gayi hai!")
```

```python
from tensorflow.keras.layers import Conv1D, MaxPooling1D, SimpleRNN, Dense, Flatten
from sklearn.metrics import accuracy_score

# --- CNN Model Implementation ---
print("--- Training CNN Model ---")
cnn_model = Sequential([
    Conv1D(filters=32, kernel_size=3, activation='relu', input_shape=(X_train.shape[1], 1)),
    MaxPooling1D(pool_size=2),
    Flatten(),
    Dense(1, activation='sigmoid')
])

cnn_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
cnn_history = cnn_model.fit(X_train_reshaped, y_train, epochs=10, batch_size=30, validation_data=(X_test_reshaped, y_test))

# --- RNN Model Implementation ---
print("\n--- Training RNN Model ---")
rnn_model = Sequential([
    SimpleRNN(32, input_shape=(X_train.shape[1], 1)),
    Dense(1, activation='sigmoid')
])

rnn_model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])
rnn_history = rnn_model.fit(X_train_reshaped, y_train, epochs=20, batch_size=40, validation_data=(X_test_reshaped, y_test))

# --- Predictions & Evaluation ---
cnn_pred_prob = cnn_model.predict(X_test_reshaped)
cnn_pred = (cnn_pred_prob > 0.5).astype(int)

rnn_pred_prob = rnn_model.predict(X_test_reshaped)
rnn_pred = (rnn_pred_prob > 0.5).astype(int)

cnn_acc = accuracy_score(y_test, cnn_pred)
rnn_acc = accuracy_score(y_test, rnn_pred)

print(f"\nCNN Test Accuracy: {cnn_acc * 100:.2f}%")
print(f"RNN Test Accuracy: {rnn_acc * 100:.2f}%")
```

```python
import json
from transformers import AutoTokenizer

# 1. JSON file load karein
with open('exercise_text.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# 2. Data ko pure text (string) mein convert karne ka safe tareeqa
if isinstance(data, dict):
    # Agar dictionary hai toh values ko mila kar string bana lein
    text_data = " ".join([str(val) for val in data.values()])
elif isinstance(data, list):
    # Agar list hai toh elements ko mila lein
    text_data = " ".join([str(item) for item in data])
else:
    text_data = str(data)

print("Extracted Text Sample:", text_data[:200])

# 3. Hugging Face Tokenizer load karein
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

# 4. Text ko tokenize karein
tokens = tokenizer.encode(text_data)

print("\n--- Tokenization Results ---")
print("Total Tokens:", len(tokens))
print("Token IDs (First 20):", tokens[:20])
```
